```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


regional_sales = sales_data_df.groupby('Region')['Sales_Amount'].sum().reset_index()
plt.figure(figsize=(8, 4))
plt.bar(regional_sales['Region'], regional_sales['Sales_Amount'], color="green") 
plt.title("Total Sales Amount by Region")
plt.xlabel("Region")
plt.ylabel("Sales Amount ($)")
plt.grid(axis='y', linestyle='--', alpha=0.7)  
plt.show()

```

```python

df_sorted = sales_data_df.sort_values('Sale_Date')
plt.figure(figsize=(12, 5))
plt.plot(df_sorted['Sale_Date'], df_sorted['Sales_Amount'].cumsum(), color='blue', linewidth=2)
plt.title("Cumulative Revenue Growth", fontsize=14)
plt.xlabel("Sale Date")
plt.ylabel("Total Lifetime Sales ($)")
plt.show()
```

```python

counts_df = pd.crosstab(sales_data_df['Customer_Type'], sales_data_df['Payment_Method'])
counts_df.plot(kind='bar', figsize=(10, 5), width=0.8)
plt.title("Transaction Counts by Customer Type and Payment Method", fontsize=14, fontweight='bold')
plt.xlabel("Customer Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)  
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title="Payment Method")
plt.tight_layout()
plt.show()
```

```python

# Boxplot automatically splits the raw rows by category without collapsing them
sns.boxplot(
    data=sales_data_df, 
    x='Product_Category', 
    y='Quantity_Sold', 
    palette='Set3'
)

plt.title("Distribution of Order Quantities by Product Category", fontsize=14, fontweight='bold')
plt.xlabel("Product Category")
plt.ylabel("Quantity per Transaction")
plt.xticks(rotation=30)
plt.grid(axis='y', linestyle=':', alpha=0.5)

plt.tight_layout()
plt.show()
```
